# Variables:

- **PassengerId**: Unique identifier for each passenger
- **Survived**: Binary indicator of survival (0 = did not survive, 1 = survived)
- **Pclass**: Ticket class (1 = first class, 2 = second class, 3 = third class)
- **Name**: Full name of the passenger
- **Sex**: Gender of the passenger (male/female)
- **Age**: Age of the passenger in years
- **SibSp**: Number of siblings/spouses aboard the Titanic
- **Parch**: Number of parents/children aboard the Titanic
- **Ticket**: Ticket number
- **Fare**: Passenger fare paid
- **Cabin**: Cabin number
- **Embarked**: Port of embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)

# Data clining

## 1. Filling in missing data

```python
df.isnull().sum()
```

```python
# ------------------------------ there are two common ways to handle missing values: fill them or remove them. ----------------------------

# 1 - this is the most common approach. Using iloc, which selects rows and columns by position.
# In this project, the following condition was applied: df.loc[condition, column_name] = replacement_value

df.loc[df.Embarked.isnull(), 'Embarked'] = 'S'

# 2 - another approach is using fillna to populate missing values. The name already says it all.
# Sintaxe: df['col'] = df['col].fillna(valor, inplace = True)

# df.Embarked = df['Embarked'].fillna("S", inplace=True)
```

```python
df.Embarked.isnull().sum() # confirming whether the Embarked column values have already been updated
```

```python
# Create a new column
df['Age/Pclass_mean'] = df.groupby(['Pclass', 'Sex'])['Age'].transform('median')
df
```

```python
df['Age_check2'] = df.Age.fillna(df.groupby(['Pclass', 'Sex'])['Age'].transform('median'))
df
```

```python
df.Age = df.Age_check2
```

```python
df.drop(['Age_check2','Age/Pclass_mean'], axis = 1, inplace = True)
```
