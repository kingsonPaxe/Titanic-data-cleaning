# About this file

The sinking of the Titanic is one of the most infamous shipwrecks in history.

On April 15, 1912, during her maiden voyage, the widely considered “unsinkable” RMS Titanic sank after colliding with an iceberg. Unfortunately, there weren’t enough lifeboats for everyone on board, resulting in the death of 1502 out of 2224 passengers and crew.

While there was some element of luck involved in surviving, it seems some groups of people were more likely to survive than others.

link: [Titanic](https://www.kaggle.com/competitions/titanic/)

# Variables:

- **PassengerId**: Unique identifier for each passenger
- **Survived**: Binary indicator of survival (0 = did not survive, 1 = survived)
- **Pclass**: Ticket class (1 = first class, 2 = second class, 3 = third class)
- **Name**: Full name of the passenger
- **Sex**: Gender of the passenger (male/female)
- **Age**: Age of the passenger in years
- **SibSp**: Number of siblings/spouses aboard the Titanic
- **Parch**: Number of parents/children aboard the Titanic
- **Ticket**: Ticket number
- **Fare**: Passenger fare paid
- **Cabin**: Cabin number
- **Embarked**: Port of embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)

# Data clining

## 1. Filling in missing data

```python
df.isnull().sum()
```

```python
# ------------------------------ existem duas formas de elimindar um dados nao nulo, ou seja preencher dados faltantes. ----------------------------

# 1 -  forma e a mais usada. Usando o iloc que é um indexador que seleciona um conjunto de linhas e colunas através de rótulos.
# Neste projecto foi aplicado a seguinte condicao: df.loc[condicao, nome_coluna] = valor_Substituto

df.loc[df.Embarked.isnull(), 'Embarked'] = 'S'

#2 - forma usando o fillna que consiste em prenhecher dados faltantes. O nome já diz tudo.
# Sintaxe: df['col'] = df['col].fillna(valor, inplace = True)

# df.Embarked = df['Embarked'].fillna("S", inplace=True)
```

```python
df.Embarked.isnull().sum() # confirmando se os dados da coluna Embarked já foram alterado
```

```python
# vendo dados faltantes novamente
df.isnull().sum()
```
